```python
from TradingviewData import TradingViewData,Interval

request = TradingViewData()
```

```python
request.search('METAL','MCX')
```

```python
nifty_data = request.get_hist(symbol='NIFTY',exchange='NSE',interval=Interval.hour_1,n_bars=1000)

nifty_data
```

```python
nifty_futures = request.get_hist(symbol='NIFTY',exchange='NSE',interval=Interval.hour_1,n_bars=1000,fut_contract=1)

nifty_futures
```

```python
relience_data = request.get_hist(symbol='RELIANCE',exchange='NSE',interval=Interval.min_5,n_bars=5000)

relience_data
```

```python
# MCX data requires a paid TradingView subscription, so use an NSE index
# for this example instead.
banknifty_data = request.get_hist(symbol='BANKNIFTY',exchange='NSE',interval=Interval.hour_1,n_bars=5000)

banknifty_data
```

```python
nifty_data.close.plot(figsize=(15,10),use_index=False)
```
